```python
import tensorflow as tf
import joblib
import numpy as np
import pickle
import os
from google.colab import drive
from pathlib import Path
from tensorflow.keras.models import Model
from tensorflow.keras.applications import InceptionResNetV2
```

```python
# Create directory for saving weights
os.makedirs('model_weights', exist_ok=True)
```

```python
drive.mount('/content/drive', force_remount = True)
```

```python
print("Loading Gender Classification Model...")
# Load the gender classification model
gender_model = tf.keras.models.load_model("/content/drive/My Drive/my_model.keras")

# Save model architecture as JSON
model_json = gender_model.to_json()
with open("model_weights/gender_model_architecture.json", "w") as json_file:
    json_file.write(model_json)

# Save model weights
gender_model.save_weights("model_weights/gender_model.weights.h5")

print("Gender model architecture and weights saved!")
```

```python
print("Loading Face Recognition Model Components...")

# Load KNN classifier
knn_model = joblib.load("/content/drive/My Drive/knn_face_recognition_model.joblib")

# Load Label Encoder
label_encoder = joblib.load("/content/drive/My Drive/label_encoder.joblib")

# Save KNN model parameters
knn_params = {
    'n_neighbors': knn_model.n_neighbors,
    'metric': knn_model.metric,
    'algorithm': knn_model.algorithm,
    'weights': knn_model.weights,
    'leaf_size': knn_model.leaf_size,
    'p': knn_model.p,
    'metric_params': knn_model.metric_params,
    'n_jobs': knn_model.n_jobs
}

# Save training data (X_train, y_train) that KNN uses for predictions
X_train = knn_model._fit_X
y_train = knn_model._y

# Save KNN components
np.save("model_weights/knn_X_train.npy", X_train)
np.save("model_weights/knn_y_train.npy", y_train)

with open("model_weights/knn_params.pkl", "wb") as f:
    pickle.dump(knn_params, f)

# Save Label Encoder
with open("model_weights/label_encoder.pkl", "wb") as f:
    pickle.dump(label_encoder, f)

print("Face recognition model components saved!")
```

```python
inception_config = {
    'model_name': 'InceptionResNetV2',
    'include_top': False,
    'pooling': 'avg',
    'input_shape': (299, 299, 3),
    'weights': 'imagenet'
}

with open("model_weights/inception_config.pkl", "wb") as f:
    pickle.dump(inception_config, f)

print("Inception ResNet V2 configuration saved!")
```
